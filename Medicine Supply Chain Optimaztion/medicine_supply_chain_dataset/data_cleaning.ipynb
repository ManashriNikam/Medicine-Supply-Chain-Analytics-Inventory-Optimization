Fix purchase_price using medicines.csv

```python
import pandas as pd

# Load datasets
po = pd.read_csv('purchase_orders.csv')
med = pd.read_csv('medicines.csv')

# Merge purchase_orders with medicines to bring in default unit_cost
po = po.merge(med[['medicine_id', 'unit_cost']], on='medicine_id', how='left')

# Fill missing purchase_price values with unit_cost from medicines
po['purchase_price'] = po['purchase_price'].fillna(po['unit_cost'])

# Drop the temporary unit_cost column
po.drop(columns=['unit_cost'], inplace=True)
```

Fix actual_delivery_date

```python
# Leave as None (NaN in pandas) so SQL/Dashboards treat them as NULL dates
po['actual_delivery_date'] = po['actual_delivery_date'].where(po['actual_delivery_date'].notnull(), None)

# Save the cleaned dataframe
po.to_csv('purchase_orders_cleaned.csv', index=False)
```
